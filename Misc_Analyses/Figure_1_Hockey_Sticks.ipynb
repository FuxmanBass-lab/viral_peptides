```python
#Need to load our data, of course...
#!
import pandas as pd 

Full_DF = pd.read_csv("VTs_withprops_SLiMs.csv", header=0)

#lets see them rows n columns:

print(Full_DF.head())
```

```python
#lets see them sticks

import seaborn as sns 

#need to generate rank

#first, clean all things that have no FC for RITA...

Full_DF = Full_DF.dropna(subset=['FC_RITA'])



#new col for the directionality of the LFC....

#little function to examine the significance row, this was't there before so we need to see
#basically based on significance and then the direction of the effect (LFC) the effect 
#that this peptide had at the cellular death response to rita, which we will 
#then used to label the markers later in the hockey stick plot..


def test_effect(row):
    
    if row['Sig_RITA'] == 'No':
        return 'None' 
    elif row['FC_RITA'] > 0:
        return 'Protective'
    else:
        return 'Deleterious'
    
Full_DF['Effect'] = Full_DF.apply(lambda row: test_effect(row), axis=1)

#for later visualization: specify size in a new row based upon the effect...


def assign_size(row):
    if row['Effect'] == 'None':
        return 45
    elif row['Effect'] == 'Protective':
        return 50
    else:
        #this is where it would be 'deleterious'
        return 50

Full_DF['Size'] = Full_DF.apply(lambda row: assign_size(row), axis=1)

#great now we use the 'rank' fn from pandas

#sort based on size, ascending 

Full_DF = Full_DF.sort_values(by='Size', ascending=True)

Full_DF['RITA_LFC_Rank'] = Full_DF['FC_RITA'].rank()


#lets see them ranks

print(Full_DF.head())
```

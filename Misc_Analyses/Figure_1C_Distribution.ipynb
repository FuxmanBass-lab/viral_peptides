### Goal: Visualize the distribution of the quantity of peptides per protein, for all the proteins in the screen. 

```python
#!
#imports

import seaborn as sns 
import matplotlib.pyplot as plt 
import pandas as pd 
import numpy as np 

from matplotlib import font_manager

from matplotlib import rcParams
rcParams['font.sans-serif'] = "Arial"
rcParams['font.family'] = "Arial"

#grab the data...

# Load positive-selection screen data.

df_rita_pos_selection = pd.read_excel(r"RITA_and_ABT_pos_selection_screens.xlsx",sheet_name = "RITA")

df_rita_pos_selection['Dropped out of Screen'] = df_rita_pos_selection['log2FoldChange'].apply(lambda x: True if  np.isnan(x) else False)

df_rita_pos_selection['Polyprotein'] = df_rita_pos_selection['Gene_name'].apply(lambda x: True if ("poly" in x) or ("Poly" in x) else False)

df_rita_pos_selection['Negative'] = df_rita_pos_selection['tileID'].apply(lambda x: True if "N" in x else False)


df_rita_pos_selection['Positive'] = df_rita_pos_selection['tileID'].apply(lambda x: True if "P" in x else False)




df_rita_pos_selection['Human Protein'] = df_rita_pos_selection['tileID'].apply(lambda x: True if 'H' in x else False)



def test_effect(row):
    
    if row['sig'] == 'No':
        return 'None' 
    elif row['log2FoldChange'] > 0:
        return 'Protective'
    else:
        return 'Deleterious'
    
df_rita_pos_selection['Effect'] = df_rita_pos_selection.apply(lambda row: test_effect(row), axis=1)

variables = ['Dropped out of Screen', "Negative","Human Protein",  "Positive", 'Polyprotein']


#merge IDs...

def label_protein_id(row):
    if pd.isnull(row['Uniprot_id']):
        return row['NCBI_id']
    else:
        return row['Uniprot_id']

df_rita_pos_selection['Protein ID'] = df_rita_pos_selection.apply(lambda row: label_protein_id(row), axis=1)

```

```python
df_nopoly = df_rita_pos_selection[(df_rita_pos_selection['Polyprotein']==False) & (df_rita_pos_selection['Human Protein']==False)]



nopoly_counts = df_nopoly['Protein ID'].value_counts()

sns.histplot(nopoly_counts, bins=120, color='#EF345C')

plt.ylim(0,140)
plt.xlim(0,140)
sns.despine(offset=10, trim=True)

plt.ylabel("Count")



plt.xlabel("Peptides / Protein")

plt.title("Distribution (no poly, no human) (w/ pos, w/ non-dropout)")

plt.show()

sns.histplot(nopoly_counts, bins=120, color='#EF345C',element='poly')


sns.despine(offset=10, trim=True)

plt.ylabel("Density")



plt.xlabel("Peptides / Protein")

plt.title("Distribution (no poly, no human) (w/ pos, w/ non-dropout)")


plt.show()

sns.histplot(nopoly_counts, bins=120, color='#EF345C',element='step')


sns.despine(offset=10, trim=True)

plt.ylabel("Density")



plt.xlabel("Peptides / Protein")

plt.title("Distribution (no poly, no human) (w/ pos, w/ non-dropout)")



plt.show()

print(f"Mean of distribution: {np.mean(nopoly_counts)}")
```

```python

sns.histplot(nopoly_counts, bins=120, color='#EF345C',element='step')
plt.figure(figsize=(5,5))
plt.ylim(0,140)
plt.xlim(0,140)
sns.despine(offset=0, trim=True)

plt.ylabel("Density")



plt.xlabel("Peptides / Protein")

plt.title("Distribution (no poly, no human) (w/ pos, w/ non-dropout)")
plt.tight_layout()

plt.savefig("Peptide_Per_Protein_Distribution.pdf",dpi=300)

##THis is the distribution seen in supp (it is edited a little for style in adobe illustrator.)
```
